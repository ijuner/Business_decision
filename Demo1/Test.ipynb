```python
import geopandas as gpd
import matplotlib.pyplot as plt
import pandas as pd

```

```python
# Sample Data (lat, lon, price)
data = pd.DataFrame({
    "lat": [39.9042, 31.2304, 34.0522],
    "lon": [116.4074, 121.4737, -118.2437],
    "price": [200, 350, 500]
})

# Convert DataFrame to GeoDataFrame
gdf = gpd.GeoDataFrame(data, geometry=gpd.points_from_xy(data.lon, data.lat))

# Load a world map
url = "https://www.naturalearthdata.com/http//www.naturalearthdata.com/download/110m/cultural/ne_110m_admin_0_countries.zip"
world = gpd.read_file(url)

# Plot the map
fig, ax = plt.subplots(figsize=(10, 6))
world.plot(ax=ax, color="lightgrey")

# Plot the points
scatter = ax.scatter(gdf.lon, gdf.lat, c=gdf.price, cmap="coolwarm", s=100, edgecolors="black")
plt.colorbar(scatter, label="Price ($)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Price Distribution on Map")

# Show the map
plt.show()
```
